# Matplot

```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['figure.figsize'] = [10,3]
plt.rcParams['figure.dpi'] = 200
```

```python
data = pd.read_csv('../01 Datasets\data\IMDB-Movie-Data.csv', nrows=60)
```

```python
data.head()
```

```python
plt.figure()
plt.plot(data['Rank'], data['Rating'])
plt.title('Rank vs Rating')
plt.xlabel('Rank')
plt.ylabel('Rating')
plt.show()
```

```python
plt.hist(data['Rating'], bins=20, edgecolor='white', color = 'orange')
plt.title('Histogram Plot')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```python
plt.scatter(data['Votes'], data['Rating'],  c = 'red')
plt.title('Scatter Plot')
plt.xlabel('Votes')
plt.ylabel('Rating')
plt.show()
```

```python
plt.boxplot(data['Rating'])
plt.show()
```

```python
data.columns
```

```python
geners = data.groupby('Genre')['Rating'].mean().sort_values(ascending = False)
plt.bar(geners.index[:10], geners.values[:10], color = 'orange')
plt.title('Rating vs Genre')
plt.xlabel('Genre')
plt.xticks(rotation=90)
plt.ylabel('Rating')
plt.show()
```

# Seaborn

```python
import seaborn as sns
```

```python
data = sns.load_dataset('iris')
data.head()
```

```python
sns.histplot(x = 'sepal_length', data = data, kde = True, bins = 20, hue='species')
plt.xlabel('sepal length in cm')
plt.ylabel('Count')
plt.title('Histogram Iris dataset')
plt.show()
```

```python
sns.pairplot(data, hue='species', height=1.7, aspect=2)
```

```python
sns.boxplot(x='species', y = 'petal_length', data = data)
```

```python
sns.swarmplot(x='species', y = 'petal_length', data = data)
```

```python
sns.boxenplot(x='species', y = 'petal_length', data = data)
```
